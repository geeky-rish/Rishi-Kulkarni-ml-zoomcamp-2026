Cell 1 — Imports and environment

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Python environment ready")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

Cell 2 — Download and load the dataset

```python
url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"

df = pd.read_csv(url)

print("Dataset shape:", df.shape)
df.head()
```

Cell 3 — Inspect the dataset

```python
print("Columns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

Cell 4 — Q1: Pandas version

```python
print("Q1 — Pandas version")
print(pd.__version__)
```

Cell 5 — Q2: Number of records

```python
record_count = len(df)

print("Q2 — Number of records:", record_count)
```

Cell 6 — Q3: Number of fuel types

```python
fuel_types = df["fuel_type"].nunique()

print("Q3 — Number of fuel types:", fuel_types)
print("\nFuel types:")
print(df["fuel_type"].unique())
```

Cell 7 — Q4: Columns with missing values

```python
missing_by_column = df.isnull().sum()

columns_with_missing = (missing_by_column > 0).sum()

print("Q4 — Number of columns with missing values:", columns_with_missing)

print("\nMissing values by column:")
print(missing_by_column[missing_by_column > 0])
```

Cell 8 — Q5: Maximum fuel efficiency for Asia

```python
asia = df[df["origin"] == "Asia"]

max_fuel_efficiency_asia = asia["fuel_efficiency_mpg"].max()

print("Q5 — Maximum fuel efficiency for cars from Asia:",
      max_fuel_efficiency_asia)
```

Cell 9 — Q6: Median horsepower

```python
# Step 1: Median before filling missing values
median_before = df["horsepower"].median()

# Step 2: Most frequent horsepower value
most_frequent_horsepower = df["horsepower"].mode()[0]

# Step 3: Fill missing horsepower values
df_filled = df.copy()

df_filled["horsepower"] = df_filled["horsepower"].fillna(
    most_frequent_horsepower
)

# Step 4: Median after filling missing values
median_after = df_filled["horsepower"].median()

print("Q6 — Horsepower")
print("-----------------------------")
print("Median before:", median_before)
print("Most frequent value:", most_frequent_horsepower)
print("Median after:", median_after)

if median_after > median_before:
    result = "Yes, it increased"
elif median_after < median_before:
    result = "Yes, it decreased"
else:
    result = "No"

print("Answer:", result)
```

Cell 10 — Q7: Linear algebra calculation

```python
# 1. Select all cars from Asia
asia = df[df["origin"] == "Asia"]

# 2. Select vehicle_weight and model_year
X_df = asia[["vehicle_weight", "model_year"]]

# 3. Select the first 7 values
X_df = X_df.head(7)

# 4. Get the underlying NumPy array
X = X_df.values

print("X:")
print(X)

# 5. Compute X^T X
XTX = X.T @ X

print("\nX^T X:")
print(XTX)

# 6. Invert X^T X
XTX_inv = np.linalg.inv(XTX)

print("\nInverse of X^T X:")
print(XTX_inv)

# 7. Create y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Compute w
w = XTX_inv @ X.T @ y

print("\nw:")
print(w)

# 9. Sum all elements
answer_q7 = w.sum()

print("\nQ7 — Sum of elements of w:", answer_q7)
print("Rounded answer:", round(answer_q7, 3))
```
